```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

# 1. Generate a synthetic 2D dataset with 4 distinct groups
# n_samples: total data points, centers: number of actual groups, cluster_std: standard deviation of clusters

```python
X, y_true = make_blobs(
    n_samples=300, centers=4, cluster_std=0.60, random_state=42
)
```

# 2. Initialize the KMeans model
# n_clusters: the number of clusters to form (k=4)
# init: 'k-means++' selects initial cluster centers smart way to speed up convergence

```python
kmeans = KMeans(n_clusters=4, init="k-means++", random_state=42)
```

# 3. Fit the model to the data and predict the cluster assignments

```python
y_kmeans = kmeans.fit_predict(X)
```

# 4. Get the coordinates of the final cluster centroids

```python
centroids = kmeans.cluster_centers_
```

# 5. Visualize the clusters and their centroids

```python
plt.figure(figsize=(8, 6))
```

# Plot the data points, coloring them by their predicted cluster label

```python
plt.scatter(
    X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap="viridis", label="Data Points"
)
```


# Plot the final centroids (marked with a red 'X')

```python
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c="red",
    s=200,
    alpha=0.9,
    marker="X",
    label="Centroids",
)
```

```python
plt.title("K-Means Clustering Visualization")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```
